```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Employee": ["Aarav", "Meera", "Kabir", "Diya", "Arjun",
                 "Anaya", "Vihaan", "Ishita", "Rohan", "Sara"],
    "Department": ["Sales", "Marketing", "Sales", "HR", "Marketing",
                   "Sales", "HR", "Marketing", "Sales", "HR"],
    "Experience": [1, 3, 5, 2, 7, 4, 6, 8, 9, 3],
    "Salary": [25000, 35000, 48000, 28000, 65000,
               42000, 55000, 72000, 80000, 32000],
    "Performance": [65, 78, 82, 70, 88, 80, 85, 92, 90, 74]
}
df=pd.DataFrame(data)
b=df.groupby("Department")["Performance"].mean()
color=["blue","gold","black"]
plt.bar(b.index,b.values,color=color)
plt.title("Performance score of departments")
plt.xlabel("Department")
plt.ylabel("Performance score")
plt.grid(axis="y",alpha=0.4)
plt.show()
```

```python
#Ivestigation
import pandas as pd 
import numpy as np

df=pd.read_csv("retail_orders_messy.csv")
print(df.shape)
print(df.columns)
print(df.head())
print(df.tail())
print(df.dtypes)
print(df.isnull().sum())
print(df.duplicated().sum())
print(df[df.duplicated(keep=False)])
print(df[df.isnull().any(axis=1)])
print(df.iloc[[8,40]])
missing=df.isnull().stack()
print(missing[missing])
print(df.describe(include="all"))
print(df[df["quantity"]<0])
print(df[df["unit_price"]<0])
print(df[df["unit_price"]>10000])
print(df[df["order_id"].duplicated(keep=False)].sort_values("order_id"))
print(df["city"].unique())
print(df["category"].unique())
print(df["payment_method"].unique())
print(df["status"].unique())
print(df[df["customer_id"]==""])
print(df["city"].value_counts(dropna=False))
print(df["category"].value_counts(dropna=False))
print(df["status"].value_counts(dropna=False))
print(df["order_date"].unique())
print(df[df["customer_id"]==""])
print(df.iloc[[13,30]].T)
for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].unique())
print(df[df["order_date"].str.contains(r"/|not available",case=False,na=False)])
for col in df.select_dtypes(include="str").columns:
    print(col,df[col].astype("string").str.strip().str.lower().isin(["","na","n/a","null","none","not available"]).sum())
```

```python
#cleaning
df["city"]=df["city"].str.strip().str.title()
print(df["city"].unique())
df["category"]=df["category"].str.strip().str.title()
print(df["category"].unique())
df["payment_method"]=df["payment_method"].str.strip().str.title()
print(df["payment_method"].unique())
df["status"]=df["status"].str.strip().str.title()
print(df["status"].unique())
df_clean=df.copy()
df_clean.isnull().sum()
print(df_clean[df_clean.isnull().any(axis=1)])
print(df_clean.iloc[24:29][["order_id","customer_id","city","product"]])
df_clean.loc[27,"customer_id"]="CUST-028"
print(df_clean.loc[27,"customer_id"])
print(df_clean[df_clean["city"].isnull()])
print(df_clean["city"].value_counts())
df_clean["city"]=df_clean["city"].fillna(df_clean["city"].mode()[0])
print(df_clean.isnull().sum())
print(df_clean[df_clean["order_date"].str.contains(r"/|not available",case=False,na=False)])
df_clean.loc[df_clean["order_id"]=="ORD-1019","order_date"]="2026-01-21"
print(df_clean.loc[0:5,["order_id","order_date"]])
df_clean.loc[df_clean["order_id"]=="ORD-1002","order_date"]="2026-01-04"
print(df_clean[df_clean["order_id"].isin(["ORD-1002","ORD-1019"])][["order_id","order_date"]])

```

```python
print(df_clean[df_clean["order_date"]=="not available"])
df_clean.loc[df_clean["order_date"]=="not available","order_date"]=pd.NA
print(df_clean[df_clean["order_date"].isnull()][["order_id","order_date"]])
df_clean["order_date"]=pd.to_datetime(df_clean["order_date"],format="%Y-%m-%d",errors="coerce")
print(df_clean["order_date"].dtype)
print(df_clean["order_date"].isnull().sum())
df_clean[df_clean.duplicated(keep=False)].sort_values("order_id")
df_clean=df_clean.drop_duplicates()
print(df_clean.shape)
print(df_clean.duplicated().sum())
q=df_clean[(df_clean["quantity"]<0)|(df_clean["unit_price"]<0)]
print(q)
df_clean.loc[df_clean["unit_price"]==-1799,"unit_price"]=1799
df_clean.loc[df_clean["quantity"]==-1,"quantity"]=1
a=df_clean[(df_clean["quantity"]<0)|(df_clean["unit_price"]<0)]
print(a)
```
